# Introduction to Ontological Agents
This notebook demonstrates an ontological agent system with three states: **Ego**, **Jaz**, and **Enotnost**.

```python
class Agent:
    def __init__(self, name, state='Ego', tension=0.0, emotion=0.0):
        self.name = name
        self.state = state
        self.tension = tension
        self.emotion = emotion
    def update_emotion(self, value, kind=None):
        self.emotion = value
        self.tension += value
    def transition(self):
        if self.tension > 0.7 or self.emotion > 0.5:
            if self.state == 'Ego':
                self.state = 'Jaz'
            elif self.state == 'Jaz':
                self.state = 'Enotnost'
        return self.state
```

## Logic and Thresholds
State transitions occur when **tension > 0.7** or **emotion > 0.5**.

```python
agent = Agent('Test')
agent.update_emotion(0.8, 'jeza')
agent.transition()
```
