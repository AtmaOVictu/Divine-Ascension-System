# Demonstration of Divine Agents
This notebook shows how to create agents, run a philosophical dialogue and inspect their transitions.

```python
from divine_agents.agent import Agent
from divine_agents.dialogue import philosophical_dialogue
```

## Create two agents
Agent **Socrates** begins in *Ego* while **Plato** starts in *Jaz*.

```python
agent1 = Agent('Socrates')
agent2 = Agent('Plato', state='Jaz')
print(f'{agent1.name} initial state: {agent1.state}')
print(f'{agent2.name} initial state: {agent2.state}')
```

## Invoke a philosophical dialogue
The conversation increases emotion and tension, leading to transitions which are logged below.

```python
lines = philosophical_dialogue(agent1, agent2)
for line in lines:
    print(line)
```

## Transition history
After the dialogue we can inspect how each agent's state evolved.

```python
print('Socrates history:', agent1.history)
print('Plato history:', agent2.history)
```
