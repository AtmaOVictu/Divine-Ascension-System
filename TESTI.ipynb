# Introduction to Ontological Agents
This notebook demonstrates an ontological agent system with three states: Ego, Jaz, and Enotnost.

## Logic and Thresholds for Transitions
Agents transition from Ego to Jaz or Jaz to Enotnost when tension > 0.7 or emotion > 0.5.

```python
from divine_agents.agent import Agent

agent = Agent('A1')
print('Initial state:', agent.state)
agent.update_emotion(0.8)
agent.transition()
print('After emotion update:', agent.state)
```

## Test Cases for Agent Transitions
Simple tests ensure that the transition logic works as expected.

```python
import unittest

class TestNotebook(unittest.TestCase):
    def test_transition(self):
        a = Agent('T')
        a.update_emotion(0.8)
        self.assertEqual(a.transition(), 'Jaz')

unittest.main(argv=[''], exit=False)
```
