# Divine Ascension Agent Demo
Demonstrates state transitions and dialogue between ontological agents.

```python
from divine_agents.agent import Agent
from divine_agents.dialogue import philosophical_dialogue

# Create two agents
ego = Agent('Ego', state='Ego')
jaz = Agent('Jaz', state='Jaz')

# Simulate tension and emotion leading to state transitions
ego.update_emotion(0.8)
jaz.update_tension(0.8)
print(f'Ego transitions to: {ego.transition()}')
print(f'Jaz transitions to: {jaz.transition()}')
```

```python
# Generate a philosophical dialogue
lines = philosophical_dialogue(ego, jaz)
for line in lines:
    print(line)
```
