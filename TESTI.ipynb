# Setup
Import the agent framework and initialize two agents.

```python
from divine_agents.agent import Agent
from divine_agents.dialogue import philosophical_dialogue

agent1 = Agent("Agent 1")
agent2 = Agent("Agent 2", state="Jaz")
agent1.state, agent2.state
```

# Dynamic Operations
Update tension and emotion to trigger transitions.

```python
agent1.update_emotion(0.6)
agent1.transition()

agent2.update_tension(0.8)
agent2.transition()

agent1.state, agent2.state
```

# Emotional Transformations
Dialogue after emotional shifts.

```python
dialogue = philosophical_dialogue(agent1, agent2)
for line in dialogue:
    print(line)
```

# Testing and Validation
Basic checks to ensure correct transitions and dialogue.

```python
assert agent1.state in Agent.STATES
assert agent1.history[-1] == agent1.state
assert any("Enotnost" in line for line in dialogue)

import unittest
from tests import test_agents
unittest.main(test_agents, argv=[''], exit=False)
```
