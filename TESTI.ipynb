# TESTI.ipynb - Example usage of the Divine-Ascension System

```python

from divine_agents.agent import Agent
from divine_agents.dialogue import philosophical_dialogue

# Create two agents with different initial states
agent1 = Agent('Agent 1', state='Ego', tension=0.6, emotion=0.4)
agent2 = Agent('Agent 2', state='Jaz', tension=0.7, emotion=0.5)

# Run a philosophical dialogue between agents
lines = philosophical_dialogue(agent1, agent2)
for line in lines:
    print(line)

# Update emotion and transition the agents
agent1.update_emotion(0.8)
agent2.update_emotion(0.6)

print('
After emotional update:')
agent1.transition()
agent2.transition()

# Display the state transitions
print(agent1.report())
print(agent2.report())
```
